```python
# 1. wap ask the user take two number print subtraction multiplication and division
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

subtraction = num1 - num2
multiplication = num1 * num2
division = num1 / num2

print("Subtraction:", subtraction)
print("Multiplication:", multiplication)
print("Division:", division)
```

```python
# 2.wap ask the user take the radius find the area of circle

radius = float(input("Enter the radius of the circle: "))

area = 3.14159 * (radius ** 2)

print("The area of the circle is:", area)
```

```python
# 3. wap ask the user take height and breadth find the area of traingle
height = float(input("Enter the height of the triangle: "))
breadth = float(input("Enter the breadth of the triangle: "))

area = 0.5 * height * breadth

print("The area of the triangle is:", area)
```

```python
# 4.wap ask the user take 3 numbers find the average
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
num3 = float(input("Enter third number: "))

average = (num1 + num2 + num3) / 3

print("The average is:", average)
```

```python
# 5. wap ask the user take the bill amount and tip amount calculate total bill
bill_amount = float(input("Enter the bill amount: "))
tip_amount = float(input("Enter the tip amount: "))

total_bill = bill_amount + tip_amount

print("The total bill is:", total_bill)
```

```python
# 6.wap ask the user take the salary, percetage of tax calculate tax amount
salary = float(input("Enter your salary: "))
tax_percent = float(input("Enter tax percentage: "))

tax_amount = (salary * tax_percent) / 100

print("The tax amount is:", tax_amount)
```
